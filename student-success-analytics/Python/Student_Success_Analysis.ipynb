```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r'[^\w]+', '_', regex=True)
    .str.strip('_')
)

df.columns.tolist()
```

```python
binary_columns = [
    'displaced',
    'educational_special_needs',
    'debtor',
    'tuition_fees_up_to_date',
    'gender',
    'scholarship_holder',
    'international'
]

for col in binary_columns:
    print(f"\n{col}")
    print(df[col].value_counts().sort_index())
```

```python
yes_no_columns = [
    'displaced',
    'educational_special_needs',
    'debtor',
    'tuition_fees_up_to_date',
    'scholarship_holder',
    'international'
]

for col in yes_no_columns:
    df[col] = df[col].map({0: 'No', 1: 'Yes'})

df[yes_no_columns].head()
```

```python
df = df.rename(columns={
    'nacionality': 'nationality'
})

df['nationality'].value_counts().sort_index()
```

```python
df['age_group'] = pd.cut(
    df['age_at_enrollment'],
    bins=[0, 20, 25, 30, 40, 100],
    labels=['20 and Under', '21-25', '26-30', '31-40', '41+']
)

df['age_group'].value_counts().sort_index()
```

```python
df['first_sem_approval_rate'] = np.where(
    df['curricular_units_1st_sem_enrolled'] > 0,
    df['curricular_units_1st_sem_approved'] /
    df['curricular_units_1st_sem_enrolled'],
    np.nan
)

df['second_sem_approval_rate'] = np.where(
    df['curricular_units_2nd_sem_enrolled'] > 0,
    df['curricular_units_2nd_sem_approved'] /
    df['curricular_units_2nd_sem_enrolled'],
    np.nan
)

df[
    ['first_sem_approval_rate', 'second_sem_approval_rate']
].describe()
```

```python
approval_by_outcome = (
    df.groupby('target', observed=True)[
        ['first_sem_approval_rate', 'second_sem_approval_rate']
    ]
    .mean()
    .round(3)
)

approval_by_outcome
```

```python
df['dropout_flag'] = (df['target'] == 'Dropout').astype(int)
df['graduate_flag'] = (df['target'] == 'Graduate').astype(int)

df[

    ['target', 'dropout_flag', 'graduate_flag']
].head(10)
```
